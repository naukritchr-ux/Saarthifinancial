# 🔮 Saarthi360 Finance Module: Predictive AI & Machine Learning Insights

Welcome to the **Saarthi360 Predictive AI Training Lab**! This Jupyter Notebook is designed to teach you step-by-step how to clean data, engineer features, train ML models, and save them for integration with your live web application.

We will build two distinct predictive models:
1. **BD Velocity & Success Predictor**: Predicts the percentage likelihood of an active deal closing and how many days it will take to close.
2. **Invoicing Leakage Propensity Model**: Predicts if an active placement is at high risk of leaking (e.g. missing invoice date or collection defaults) before it happens.

### Step 1: Import Required Libraries
We will use `pandas` and `numpy` for data manipulation, `mysql.connector` to fetch data from our live database, and `scikit-learn` to build, train, and evaluate our machine learning models.

```python
import os
import pandas as pd
import numpy as np
import mysql.connector
from datetime import datetime
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
import sklearn.metrics as metrics
import joblib

print("✅ Libraries imported successfully!")
```

### Step 2: Connect to Live MySQL Database
We will query our MySQL database `crm_db` directly to extract the training features. This ensures our AI is trained on the latest clean records.

```python
db_config = {
    'host': 'localhost',
    'user': 'root',
    'password': '',
    'database': 'crm_db'
}

try:
    conn = mysql.connector.connect(**db_config)
    print("✅ Successfully connected to live crm_db!")
except Exception as e:
    print("❌ Failed to connect to MySQL:", str(e))
```

### Step 3: Load Data and Feature Engineering
Let's fetch the data from the `enquiries` table, join it with the `invoice` details, and preprocess features:
- **Target 1 (Velocity)**: Calculated as the duration in days between `dateOfAllocation` and `bill_date` for closed deals.
- **Target 2 (Success)**: Set to `1` if status is `'closed'`, and `0` if status is `'cancelled'`.
- **Categorical Encoders**: Encode recruiter names, client companies, and positions into numeric formats.

```python
query = """
    SELECT 
        e.id, 
        e.companyName, 
        e.bdMemberName, 
        e.positionName, 
        e.placementFees, 
        e.enquiryStatus,
        e.dateOfAllocation, 
        e.bill_date, 
        e.bill_amount,
        i.amountReceived
    FROM enquiries e
    LEFT JOIN invoice i ON e.id = i.enquiry_id
"""

try:
    df = pd.read_sql(query, conn)
    print(f"📊 Loaded {len(df)} records from the database.")
except Exception as e:
    print("Loading failed, using fallback CSV loader:", str(e))
    # Fallback to local CSV if not connected to live DB in this terminal
    df = pd.read_csv('../backend/master_final_pipeline.csv')

# 1. Calculate Target Variables
df['dateOfAllocation'] = pd.to_datetime(df['Date of Allocation'] if 'Date of Allocation' in df.columns else df['dateOfAllocation'])
df['bill_date'] = pd.to_datetime(df['Bill Date'] if 'Bill Date' in df.columns else df['bill_date'])
df['days_to_close'] = (df['bill_date'] - df['dateOfAllocation']).dt.days

# Success Target: closed=1, cancelled/inprogress=0
status_col = 'Client Status' if 'Client Status' in df.columns else 'enquiryStatus'
df['is_success'] = df[status_col].apply(lambda x: 1 if str(x).strip().lower() == 'closed' else 0)

# 2. Leakage Label: bill number exists but bill date is missing
bill_no_col = 'Bill Number' if 'Bill Number' in df.columns else 'bill_no'
df['is_leak'] = 0
df.loc[df['bill_date'].isnull() & df[bill_no_col].notnull(), 'is_leak'] = 1

# 3. Fill missing values
placement_col = 'Placement Fees' if 'Placement Fees' in df.columns else 'placementFees'
bill_amt_col = 'Service Charges' if 'Service Charges' in df.columns else 'bill_amount'
amt_rec_col = 'Amount Received' if 'Amount Received' in df.columns else 'amountReceived'

df['placementFees'] = pd.to_numeric(df[placement_col], errors='coerce').fillna(0.0)
df['bill_amount'] = pd.to_numeric(df[bill_amt_col], errors='coerce').fillna(0.0)
df['amountReceived'] = pd.to_numeric(df[amt_rec_col], errors='coerce').fillna(0.0)

# Encode categories
le_company = LabelEncoder()
company_col = 'Company Name' if 'Company Name' in df.columns else 'companyName'
df['company_encoded'] = le_company.fit_transform(df[company_col].fillna('Unknown'))

le_bd = LabelEncoder()
bd_col = 'BD Member' if 'BD Member' in df.columns else 'bdMemberName'
df['bd_encoded'] = le_bd.fit_transform(df[bd_col].fillna('Unknown'))

pos_col = 'Position Name' if 'Position Name' in df.columns else 'positionName'
df['position_len'] = df[pos_col].apply(lambda x: len(str(x)) if x else 0)

df.head()
```

### Step 4: Model 1 Training - BD Velocity Forecasting
We train a `RandomForestRegressor` to estimate how many days a placement takes to close based on its characteristics.

```python
# Filter to closed/billed placements only for velocity training
df_closed = df[df['days_to_close'].notnull() & (df['days_to_close'] >= 0)]

X_vel = df_closed[['company_encoded', 'bd_encoded', 'placementFees', 'bill_amount', 'position_len']]
y_vel = df_closed['days_to_close']

X_train, X_test, y_train, y_test = train_test_split(X_vel, y_vel, test_size=0.2, random_state=42)

vel_model = RandomForestRegressor(n_estimators=100, random_state=42)
vel_model.fit(X_train, y_train)

# Evaluate
y_pred = vel_model.predict(X_test)
mae = metrics.mean_absolute_error(y_test, y_pred)
r2 = metrics.r2_score(y_test, y_pred)

print(f"📈 BD Velocity Model trained successfully!")
print(f"   - Mean Absolute Error (MAE): {mae:.2f} Days")
print(f"   - R² Regression Score: {r2:.3f}")
```

### Step 5: Model 2 Training - Invoicing Leakage Propensity Model
We train a `RandomForestClassifier` to estimate the probability percentage of a deal succeeding and its potential to leak.

```python
X_risk = df[['company_encoded', 'bd_encoded', 'placementFees', 'bill_amount', 'position_len']]
y_success = df['is_success']

X_train, X_test, y_train, y_test = train_test_split(X_risk, y_success, test_size=0.2, random_state=42)

success_model = RandomForestClassifier(n_estimators=100, random_state=42)
success_model.fit(X_train, y_train)

# Evaluate
y_pred = success_model.predict(X_test)
acc = metrics.accuracy_score(y_test, y_pred)
print(f"🛡️ Success Predictor Model trained successfully!")
print(f"   - Classification Accuracy: {acc * 100:.2f}%")
```

### Step 6: Serialize Models for Dashboard Integration
We save the models and label encoders as Joblib serialization files. The Flask backend (`app.py`) will load these files to serve real-time predictions.

```python
# Export files
os.makedirs('../backend/models', exist_ok=True)
joblib.dump(vel_model, '../backend/models/velocity_model.joblib')
joblib.dump(success_model, '../backend/models/success_model.joblib')
joblib.dump(le_company, '../backend/models/le_company.joblib')
joblib.dump(le_bd, '../backend/models/le_bd.joblib')

print("💾 All models and label encoders serialized inside /backend/models/ directory!")
```
