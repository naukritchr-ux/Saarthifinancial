```python
import pandas as pd

df = pd.read_csv(r'C:\Users\ADMIN\Downloads\master_final_enquiry_sheet_with_position_name.csv')

df1= pd.DataFrame(df)
print(df1.columns)
print(df1['Service Charges'])
print(df1['Total Bill Amount'])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.plot(df1['Service Charges'],df1['Total Bill Amount'], marker='o')
plt.show()

sns.boxplot(df1['Total Bill Amount'])
plt.show()
```
